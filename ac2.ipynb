# 1. Indexació i slicing

Les seqüències (`str`, `list`, `tuple`, `range`...) permeten accedir als seus elements mitjançant índexs.

```text
seq[inici:final:pas]
```

- `inici`: inclòs
- `final`: exclòs
- `pas`: salt entre elements

Els índexs negatius recorren la seqüència des del final.

A diferència d'alguns usos habituals de JavaScript, Python permet fer *slicing* directament amb aquesta sintaxi.


````python
temperatures = [22, 21, 23, 24, 26, 25, 27, 28]

print(temperatures[0])      # primer element
print(temperatures[-1])     # últim element
print(temperatures[-3:])    # tres últims
print(temperatures[1:5])    # índex 1 fins a 4
print(temperatures[::2])    # un de cada dos
print(temperatures[::-1])   # ordre invers
````

### Exercici 2 — Slicing bidimensional

Una matriu és una llista de llistes. El slicing de la matriu i el slicing de cada fila són dues operacions diferents.

La següent expressió **no** selecciona una submatriu rectangular:

matrix[1:4][1:4]

Per què?

Construeix la submatriu 3×3 de la cantonada inferior dreta utilitzant slicing de les files.


````python
matrix = [[i + j for j in range(5)] for i in range(5)]

# Resultat esperat:
# [[2, 3, 4],
#  [3, 4, 5],
#  [4, 5, 6]]

# Escriu aquí la teva solució
result = [[matrix[i][j] for j in range(2, 5)] for i in range(2, 5)]
print(result)
````

## 3. Iterar amb `enumerate()` i `zip()`

Quan necessitem **índex + valor**, `enumerate()` evita haver de gestionar manualment un comptador.

Quan volem recórrer dues seqüències en paral·lel, `zip()` és l'eina idiomàtica (és a dir, pròpia i habitual del llenguatge Python)


````python
features = [0.1, 0.5, 0.3]

for i, feature in enumerate(features):
    print(f"Feature {i}: {feature}")

names = ["Ada", "Linus", "Guido"]
scores = [95, 87, 92]

for name, score in zip(names, scores):
    print(f"{name}: {score}")
````

## 4. Desempaquetat

Python permet assignar diversos valors d'una seqüència a diverses variables:

```python
a, b, c = [10, 20, 30]
```

També podem utilitzar `*` per capturar la resta:

```python
primer, *mig, ultim = range(6)
```

Això és especialment útil quan coneixem algunes posicions però no volem calcular índexs manualment.


````python
a, b, c = [10, 20, 30]
print(a, b, c)

primer, *mig, ultim = range(6)
print(primer)
print(mig)
print(ultim)

# Intercanvi de variables
a, b = 5, 10
a, b = b, a
print(a, b)
````

## 5. `split()`, `strip()` i `join()`

Aquest trio apareix constantment quan processem dades textuals.

- `split()` → string → llista
- `strip()` → elimina espais (o caràcters indicats) dels extrems
- `join()` → llista de strings → string


````python
text = "  Python, JavaScript, SQL  "

parts = text.strip().split(",")
parts = [part.strip() for part in parts]

print(parts)

resultat = " | ".join(parts)
print(resultat)
````

# 6. List comprehensions

Una *list comprehension* permet construir una llista a partir d'una iteració de manera compacta.

Patró bàsic:

```python
[resultat for element in seqüencia]
```

Amb condició:

```python
[resultat for element in seqüencia if condicio]
```

És l'equivalent idiomàtic d'un `for` que va acumulant resultats en una llista.


````python
# Versió tradicional
squares = []
for i in range(6):
    squares.append(i ** 2)

print(squares)

# Amb comprehension
squares = [i ** 2 for i in range(6)]
print(squares)

even_squares = [i ** 2 for i in range(10) if i % 2 == 0]
print(even_squares)
````

## 7. Comprehensions niuades

Podem tenir més d'un `for` dins d'una comprehension.

Per exemple, per aplanar:

```python
matrix = [[1, 2, 3], [4, 5, 6]]
```

podem escriure:

```python
flattened = [value for row in matrix for value in row]
```

L'ordre dels `for` correspon a l'ordre dels bucles tradicionals.


````python
matrix = [[1, 2, 3], [4, 5, 6]]

flattened = [value for row in matrix for value in row]
print(flattened)

# Mateix patró amb bucles tradicionals:
flattened = []
for row in matrix:
    for value in row:
        flattened.append(value)

print(flattened)
````

# 8. Mètodes de llistes

Alguns mètodes importants:

| Mètode | Efecte |
|---|---|
| `append(x)` | afegeix un element |
| `extend(seq)` | afegeix diversos elements |
| `insert(i, x)` | insereix en una posició |
| `pop()` | elimina i retorna un element |
| `sort()` | ordena la llista **in situ** |
| `reverse()` | inverteix la llista **in situ** |
| `index(x)` | primera posició de `x` |
| `count(x)` | nombre d'aparicions |

Un punt important: molts mètodes de llista **modifiquen la llista i retornen `None`**.


````python
values = [3, 1, 2]

values.append(4)
print(values)

values.extend([5, 6])
print(values)

removed = values.pop()
print("Eliminat:", removed)
print(values)

values.sort()
print(values)
````

# 9. Mètodes de strings i preprocessament

Quan processem dades reals, sovint cal normalitzar el text abans d'analitzar-lo.

Alguns mètodes útils:

```python
lower()
upper()
strip()
replace()
startswith()
endswith()
isdigit()
isalpha()
```

Podem combinar-los amb comprehensions i condicionals.


````python
raw = "  Python, PYTHON, python  "

clean = raw.strip().lower().replace(" ", "")
print(clean)

words = ["Python", "JavaScript", "Python"]
python_words = [word for word in words if word.lower() == "python"]
print(python_words)
````

# 10. Repte integrador — Processador de registres

Ara combinarem les tècniques de tota la pràctica.

Disposem d'un conjunt de registres textuals:

```python
raw_data = [
    "A01 | Ada | 9.5 | Python",
    "A02 | BOB | 7.0 | JavaScript",
    "A03 |  carla | 8.5 | Python",
    "A04 | David | 5.5 | JavaScript",
    "A05 | EVE | 9.0 | Python"
]
```

## Tasca

Construeix un programa que:

1. **Neteja i separa** cada registre.
2. Normalitza el nom (`strip()` + format consistent).
3. Converteix la puntuació a `float`.
4. Utilitza **desempaquetat** per separar els camps.
5. Genera una llista de registres estructurats.
6. Filtra només els registres de **Python**.
7. Ordena aquests registres de major a menor puntuació.
8. Mostra un informe numerat utilitzant `enumerate()`.

### Resultat aproximat

```text
1. Ada       -> 9.5
2. Eve       -> 9.0
3. Carla     -> 8.5
```

### Restriccions

Has d'utilitzar, com a mínim:

- `split()`
- `strip()`
- `join()` o una f-string per generar la sortida
- desempaquetat
- una *list comprehension*
- `sorted()`
- `enumerate()`

**No cal fer-ho tot en una única línia.** Prioritza la claredat.


````python
raw_data = [
    "A01 | Ada | 9.5 | Python",
    "A02 | BOB | 7.0 | JavaScript",
    "A03 |  carla | 8.5 | Python",
    "A04 | David | 5.5 | JavaScript",
    "A05 | EVE | 9.0 | Python"
]

# 1. Neteja i estructura els registres
# 2. Filtra Python
# 3. Ordena per puntuació descendent
# 4. Mostra el informe

# Escriu aquí la teva solució
registres = []
for item in raw_data:
    id_registre, nom, puntuacio, categoria = item.split("|")

    name = nom.strip().title()
    punctuation = float(puntuacio.strip())
    category = categoria.strip()

    registres.append((id_registre.strip(), name, punctuation, category))
````
